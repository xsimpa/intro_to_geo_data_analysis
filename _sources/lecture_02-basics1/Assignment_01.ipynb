## Part 3: Volcanic eruption

You are a volcanologist studying three volcanoes. Each volcano has erupted three times. Use Python to analyze the Volcanic Explosivity Index (VEI).


| Volcano   | Eruption 1 | Eruption 2 | Eruption 3 |
|-----------|------------|------------|------------|
| Volcano A | 2          | 4          | 3          |
| Volcano B | 5          | 6          | 4          |
| Volcano C | 1          | 2          | 2          |


1. Store the VEIs in a nested list.

2. Use a while or for loop to calculate the mean VEI for each volcano.

3. Use if, elif, and else to classify the mean:
- Below 2: Low explosivity rate
- From 2 to below 4: Moderate explosivity rate
- 4 or higher: High explosivity rate
3. Count how many eruptions have a VEI of 4 or higher.

4. Identify the highest VEI recorded across all three volcanoes.
